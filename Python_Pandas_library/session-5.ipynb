```python
# 1.Download a CSV file of IPL cricket players with columns for player name, team, and age, then use pandas to load it and print out all rows where the age column has missing values.


import pandas as pd

players = pd.read_csv("ipl_players.csv")

missing_age = players[players["age"].isnull()]

print("Players with Missing Age:")
print(missing_age)
```

```python
# 2.Using a dataset of food delivery orders (with columns: order_id, customer_name, delivery_rating), use pandas to drop all rows where the delivery_rating is missing and display the cleaned DataFrame.

import pandas as pd

orders = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106],
    "customer_name": [
        "Dhruv",
        "Rahul",
        "Amit",
        "Jay",
        "Karan",
        "Vivek"
    ],
    "delivery_rating": [5.0, 4.0, None, 3.5, None, 4.5]
})

cleaned_orders = orders.dropna(subset=["delivery_rating"])

print("Original DataFrame:")
print(orders)

print("\nCleaned DataFrame:")
print(cleaned_orders)
```

```python
# 3.Given a Flipkart-style product reviews dataset with some missing 'rating' values, use pandas to fill all missing ratings with the mean rating of the dataset and print the updated DataFrame.<br><br><em><strong>Hint:</strong> Use the fillna() method with the mean() function.</em>


import pandas as pd

reviews = pd.DataFrame({
    "product": [
        "Laptop",
        "Phone",
        "Headphones",
        "Keyboard",
        "Mouse",
        "Monitor"
    ],
    "rating": [4.5, 4.0, None, 3.5, 5.0, None]
})

mean_rating = reviews["rating"].mean()

reviews["rating"] = reviews["rating"].fillna(mean_rating)

print("Mean Rating:", mean_rating)
print("\nUpdated Reviews:")
print(reviews)
```

```python
# 4.You have a Spotify playlist CSV with missing 'duration' values for some songs. Replace all missing durations with the median duration of the available songs using pandas, then save the cleaned DataFrame to a new CSV file.

import pandas as pd

playlist = pd.read_csv("spotify_playlist.csv")

median_duration = playlist["duration"].median()

playlist["duration"] = playlist["duration"].fillna(median_duration)

playlist.to_csv("cleaned_spotify_playlist.csv", index=False)

print("Median Duration:", median_duration)
print("\nCleaned Playlist:")
print(playlist)

print("\nCleaned file saved as cleaned_spotify_playlist.csv")
```
