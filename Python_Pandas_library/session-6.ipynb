```python
# 1.Load a CSV file of Spotify song streams (with columns: song, artist, streams, date) into a pandas DataFrame and use drop_duplicates() to remove any duplicate song entries, keeping only the first occurrence.


import pandas as pd

df = pd.read_csv("spotify_songs.csv")

print("Before removing duplicates:")
print(df)

df = df.drop_duplicates(subset=["song"], keep="first")

print("\nAfter removing duplicates:")
print(df)
```

```python
# 2.Given a pandas DataFrame of Flipkart product reviews with columns: product_id, user_id, rating, review_text, identify and print the number of duplicate reviews (where product_id and user_id are both the same) before and after using drop_duplicates().


import pandas as pd

df = pd.DataFrame({
    "product_id": [101, 102, 101, 103, 102, 104],
    "user_id": [1, 2, 1, 3, 2, 4],
    "rating": [5, 4, 5, 3, 4, 5],
    "review_text": [
        "Excellent",
        "Good product",
        "Excellent",
        "Average",
        "Good product",
        "Very good"
    ]
})

duplicate_count_before = df.duplicated(
    subset=["product_id", "user_id"]
).sum()

df_clean = df.drop_duplicates(
    subset=["product_id", "user_id"],
    keep="first"
)

duplicate_count_after = df_clean.duplicated(
    subset=["product_id", "user_id"]
).sum()

print("Duplicate reviews before:", duplicate_count_before)
print("Duplicate reviews after:", duplicate_count_after)
```

```python
# 3.Use the IQR method to detect outliers in the 'order_amount' column of a Zomato orders DataFrame and print the order IDs that are considered outliers.<br><br><em><strong>Hint:</strong> Calculate Q1, Q3, and IQR, then filter orders outside the [Q1 - 1.5*IQR, Q3 + 1.5*IQR] range.</em>


import pandas as pd

df = pd.DataFrame({
    "order_id": [1, 2, 3, 4, 5, 6, 7, 8],
    "order_amount": [250, 300, 450, 500, 550, 600, 700, 5000]
})

Q1 = df["order_amount"].quantile(0.25)
Q3 = df["order_amount"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[
    (df["order_amount"] < lower_limit) |
    (df["order_amount"] > upper_limit)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)

print("\nOutlier Order IDs:")
print(outliers["order_id"].tolist())
```

```python
# 4.For a DataFrame containing daily IPL match ticket sales (columns: date, tickets_sold), calculate the Z-score for each day's tickets_sold, and print the dates where the absolute Z-score is greater than 2.<br><br><em><strong>Constraint:</strong> Use pandas and scipy.stats.zscore for calculations.</em>


import pandas as pd
from scipy.stats import zscore

df = pd.DataFrame({
    "date": [
        "2026-01-01",
        "2026-01-02",
        "2026-01-03",
        "2026-01-04",
        "2026-01-05",
        "2026-01-06",
        "2026-01-07"
    ],
    "tickets_sold": [500, 520, 510, 530, 515, 525, 2000]
})

df["z_score"] = zscore(df["tickets_sold"])

outliers = df[abs(df["z_score"]) > 2]

print("Outlier dates:")
print(outliers[["date", "tickets_sold", "z_score"]])
```

```python
# 5.Use ChatGPT or Copilot to generate a pandas function that finds and returns all outlier transaction amounts in a Paytm wallet DataFrame using the Z-score method, then test the function with sample data.


import pandas as pd
from scipy.stats import zscore

def find_outliers(df, column):
    df = df.copy()
    df["z_score"] = zscore(df[column])
    return df[abs(df["z_score"]) > 2]

df = pd.DataFrame({
    "transaction_id": [101, 102, 103, 104, 105, 106, 107],
    "amount": [200, 250, 300, 220, 280, 260, 5000]
})

outliers = find_outliers(df, "amount")

print("Outlier transactions:")
print(outliers)
```
