```python
# 2.Perform univariate analysis on the 'runs' column: calculate and print the mean, median, mode, minimum, and maximum runs scored by batsmen in the IPL_2023_Batsmen dataset.

import pandas as pd

df = pd.read_csv("IPL_2023_Batsmen.csv")

runs = df["runs"]

print("Mean:", runs.mean())
print("Median:", runs.median())
print("Mode:", runs.mode().tolist())
print("Minimum:", runs.min())
print("Maximum:", runs.max())
```

```python
# 3.Create a Pandas DataFrame for your top 5 favorite Zomato restaurants with columns: 'name', 'rating', 'votes', and 'avg_cost'. Perform bivariate analysis between 'rating' and 'votes' by calculating their correlation coefficient.

import pandas as pd

restaurants = pd.DataFrame({
    "name": [
        "Dominos",
        "McDonalds",
        "The Food Hub",
        "Spice Garden",
        "Pizza Point"
    ],
    "rating": [4.5, 4.2, 4.7, 4.0, 4.4],
    "votes": [1200, 950, 1800, 600, 1100],
    "avg_cost": [500, 400, 700, 350, 550]
})

correlation = restaurants["rating"].corr(restaurants["votes"])

print("Correlation between rating and votes:", correlation)
```
